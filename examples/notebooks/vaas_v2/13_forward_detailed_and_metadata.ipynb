This notebook explores the **detailed outputs** exposed by the VAAS pipeline.

It focuses on:

- `forward_detailed`
- model `variant`
- model `metadata`
- how to use these in research workflows

## 2. Imports

```python
from io import BytesIO
import requests
from PIL import Image

from vaas.inference.pipeline import VAASPipeline
```

## 3. Load image

```python
url = "https://raw.githubusercontent.com/OBA-Research/VAAS/main/examples/images/COCO_DF_C110B00000_00539519.jpg"
image = Image.open(BytesIO(requests.get(url).content)).convert("RGB")
image
```

## 4. Load pipeline

```python
pipeline = VAASPipeline.from_pretrained(
    repo_id="OBA-Research/vaas",
    model_variant="v2-base-df2023",
    device="cpu",
    alpha=0.5,
)
```

## 5. Standard inference

This returns the minimal output used in most workflows.

```python
result = pipeline(image)

print(result.keys())
```

## 6. Detailed inference

`forward_detailed` exposes additional structured outputs.

```python
detailed = pipeline.forward_detailed(image)

print(detailed.keys())
```

## 7. Inspect scores

```python
print("S_F:", detailed["S_F"])
print("S_P:", detailed["S_P"])
print("S_H:", detailed["S_H"])
```

## 8. Access anomaly map

```python
anomaly_map = detailed["anomaly_map"]
print(anomaly_map.shape)
```

## 9. Model variant

```python
print("Variant:", detailed["variant"])
```

## 10. Metadata inspection

This contains structured information about the model.

```python
metadata = detailed["metadata"]

for k, v in metadata.items():
    print(k, ":", v)
```

## 11. Why metadata matters

Typical uses:

- experiment tracking  
- dataset provenance  
- reproducibility  
- model comparison  

## 12. Example: logging outputs

```python
log = {
    "variant": detailed["variant"],
    "S_F": detailed["S_F"],
    "S_P": detailed["S_P"],
    "S_H": detailed["S_H"],
    "metadata": detailed["metadata"],
}

log
```
