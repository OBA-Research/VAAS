- Structural manipulation (e.g. copy-move)
- AI-generated / synthetic manipulation

## 2. Imports

```python

from io import BytesIO
import requests
from PIL import Image
from IPython.display import display, Image as IPImage

from vaas.inference.pipeline import VAASPipeline
```

## 3. Load images

```python

# Structural example
url_struct = "https://raw.githubusercontent.com/OBA-Research/VAAS/main/examples/images/vaas_dfrws_Structural_manipulated.jpg"

# AI-like / synthetic example
url_ai = "https://raw.githubusercontent.com/OBA-Research/VAAS/main/examples/images/vaas_dfrws_ai_edited.png"

img_struct = Image.open(BytesIO(requests.get(url_struct).content)).convert("RGB")
img_ai = Image.open(BytesIO(requests.get(url_ai).content)).convert("RGB")

display(img_struct)
display(img_ai)
```

## 4. Load pipeline

```python

pipeline = VAASPipeline.from_pretrained(
    repo_id="OBA-Research/vaas",
    model_variant="v2-large-df2023",
    device="cpu",
    alpha=0.5,
)
```

## 5. Run inference

```python

out_struct = pipeline(img_struct)
out_ai = pipeline(img_ai)

print("Structural:", out_struct["S_H"])
print("AI-generated:", out_ai["S_H"])
```

## 6. Generate visual outputs

```python

pipeline.visualize(img_struct, "structural.png", mode="all", threshold=0.5)
pipeline.visualize(img_ai, "ai_generated.png", mode="all", threshold=0.5)
```

## 7. Display results

```python

display(IPImage("structural.png"))
display(IPImage("ai_generated.png"))
```


## 8. Interpretation

- VAAS localises manipulated regions in both cases
- Hybrid score adapts to manipulation strength
- Works across different manipulation types
