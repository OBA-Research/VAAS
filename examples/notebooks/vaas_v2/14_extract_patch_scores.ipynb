# 14. VAAS v2 — Extract Patch Scores

This notebook demonstrates how to directly access **patch-level anomaly outputs**.

It focuses on:

- extracting patch scores
- visualising patch-level outputs independently
- using patch scores for downstream analysis

## 2. Imports

```python
from io import BytesIO
import requests
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image

from vaas.inference.pipeline import VAASPipeline
```

## 3. Load image

```python
url = "https://raw.githubusercontent.com/OBA-Research/VAAS/main/examples/images/COCO_DF_C110B00000_00539519.jpg"
image = Image.open(BytesIO(requests.get(url).content)).convert("RGB")
image
```

## 4. Load pipeline

```python
pipeline = VAASPipeline.from_pretrained(
    repo_id="OBA-Research/vaas",
    model_variant="v2-base-df2023",
    device="cpu",
    alpha=0.5,
)
```

## 5. Extract patch-level anomaly map

The anomaly map corresponds to patch-level anomaly reasoning.

```python
result = pipeline(image)
anomaly_map = result["anomaly_map"]

print("Shape:", anomaly_map.shape)
```

## 6. Visualise patch-level heatmap

```python
plt.imshow(anomaly_map, cmap="jet")
plt.colorbar()
plt.title("Patch-level anomaly map")
plt.axis("off")
plt.show()
```

## 7. Overlay heatmap on image

```python
plt.figure(figsize=(6,6))
plt.imshow(image)
plt.imshow(anomaly_map, cmap="jet", alpha=0.4)
plt.title("Overlay: Image + Patch anomaly")
plt.axis("off")
plt.show()
```

## 8. Threshold patch scores

```python
threshold = 0.5
binary_map = (anomaly_map > threshold).astype(np.uint8)

plt.imshow(binary_map, cmap="gray")
plt.title("Binary anomaly mask")
plt.axis("off")
plt.show()
```

## 9. Save outputs

```python
np.save("patch_scores.npy", anomaly_map)
np.save("binary_mask.npy", binary_map)

print("Saved patch_scores.npy and binary_mask.npy")
```
